```python
from transformers import pipeline


```

```python
# Tareas básicas
sentiment_analysis = pipeline("sentiment-analysis")
text_classification = pipeline("text-classification", model="distilbert-base-uncased-finetuned-sst-2-english")




question_answering = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
```

```python

# Iniciar pruebas
print(sentiment_analysis("I love this new AI model!"))
```

```python
Analisis de sentimiento
print(sentiment_analysis("El partido estuvo terrible"))
```

```python
#Clasificación y analisis de sentimiento
print(text_classification("This movie was mind-blowing and unforgettable!"))
print(sentiment_analysis("This movie was mind-blowing and unforgettable!"))
```

```python
#Pregunta - Respuesta
context = "El sol es la principal fuente de energía para la Tierra. Sus rayos permiten la fotosíntesis en las plantas."
question = "¿Cuál es la principal fuente de energía para la Tierra?"
print(question_answering(question=question, context=context))
```

```python
#Generación de Texto
text_generator = pipeline("text-generation", model="gpt2")
text1=(text_generator("Once upon a time", max_length=50))
print(text1)
```

```python
#Traducción automática
translator = pipeline("translation", model="Helsinki-NLP/opus-mt-es-en")
print(translator("Hola, ¿cómo estás hijo de la chingada?"))
```
