```python
import getpass
import os

if not os.environ.get("GOOGLE_API_KEY"):
  os.environ["GOOGLE_API_KEY"] = getpass.getpass("Ingresar API key de Google Gemini: ")

```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-2.0-flash", temperature=0.7)
```

```python
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationChain

#Creamos un componente que retenga conversaciones con LLMs
memory = ConversationBufferMemory(return_messages=True)

#Configuramos un chain mas complejo incluyendo el LLM
#y un componente de Memoria
# A diferencia de prompt template, los prompts son
# creados automaticamente (ver abajo) y el contexto
# es manejado automaticamente contra el LLM
conversation = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=True  # Muestra pasos intermedios
)
```

```python
# Notar en el output los roles de AI/Human
# First message
response = conversation.predict(input="Hi, I'm learning LangChain!")
print(response)

# Second message, testeando memoria de forma simple
response = conversation.predict(input="What did I just say?")
print(response)

# Third message, testeando memoria y contexto completo
response = conversation.predict(input="Summarize our conversation so far.")
print(response)
```
