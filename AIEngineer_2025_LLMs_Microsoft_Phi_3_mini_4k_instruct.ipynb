```python
from transformers import AutoModelForCausalLM, AutoTokenizer, GenerationConfig

# Cargar el modelo y el tokenizador
# Primero, se importa el modelo para el lenguaje causal (generación de texto)
# desde el repositorio de modelos preentrenados de la libreria transformers.
# El parámetro "device_map" se establece en "cuda" para usar la GPU
#si está disponible.
# El parámetro "torch_dtype" se establece en "auto" para seleccionar
# automáticamente el tipo de datos adecuado.
# El parámetro "trust_remote_code" se establece en True
# para confiar en el código remoto que estamos bajando de Hugging Face.
model = AutoModelForCausalLM.from_pretrained(
    "microsoft/Phi-3-mini-4k-instruct",
    device_map="cuda",
    torch_dtype="auto",
    trust_remote_code=True,
)

# Se carga el tokenizador asociado al modelo preentrenado.
# El tokenizador es responsable de convertir el texto
# en tokens que el modelo puede procesar.
tokenizer = AutoTokenizer.from_pretrained("microsoft/Phi-3-mini-4k-instruct")
```

```python
from transformers import pipeline

# Crear un pipeline para la generación de texto
# Un pipeline es una forma simplificada de utilizar
# modelos preentrenados en tareas específicas.
# En este caso, se crea un pipeline para la generación de texto
# utilizando el modelo y el tokenizador cargados.
# return_full_text=False indica que no se debe devolver el texto completo,
# solo la parte generada como respuesta.
# max_new_tokens=500 establece el número máximo de nuevos tokens
# que se pueden generar.
# do_sample=False indica que no se debe muestrear (usar la estrategia greedy).
# Al finalizar el proceso SIEMPRE elige el token con mayor probabilidad, no es
# al azar como normalmente. Esta estrategia optimiza la velocidad de respuesta
generation_config = GenerationConfig(use_cache=False)


generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    generation_config=generation_config, #Pass the GenerationConfig object
    return_full_text=False,
    max_new_tokens=500,
    do_sample=False
)
```

```python
# La entrada del usuario (prompt)
# Se define una lista de mensajes que actúan
# como la entrada del usuario para el modelo.
# Cada mensaje tiene un rol y un contenido.
#En este caso, el rol es "user" y el contenido es la solicitud del usuario.
messages = [
    {"role": "user", "content": "Give me a chocolate cake recipe."}
]
```

```python
# Generar la salida
# Se utiliza el pipeline de generación de texto para producir una respuesta
# basada en el input prompt.
# La salida es una lista de diccionarios,
# donde cada diccionario contiene el texto generado.
output = generator(messages)
```

```python
# Imprimir el texto generado
# Se imprime el texto generado por el modelo .
print(output[0]["generated_text"])
```
