```python
import tiktoken
```

```python

# Función para tokenizar y mapear texto a IDs de tokens
def map_text_to_token_ids(text, model="gpt-4"):
    # Cargar el tokenizador para el modelo especificado
    tokenizer = tiktoken.encoding_for_model(model)

    # Tokenizar el texto de entrada
    tokens = tokenizer.encode(text)

    # Mapear tokens a sus IDs y texto decodificado
    token_details = [{"Token ID": token, "Token Text": tokenizer.decode([token])} for token in tokens]

    return token_details
```

```python
#Costos API (input)
OpenAI_API_costs1M=2.5
OpenAI_API_costs={"1m":round(OpenAI_API_costs1M/1000000,10)}
```

```python
# Texto de entrada de ejemplo
input_text = "no vives de ensaladas pero si de un buen asado con cuero"
```

```python
# Tokenizar el texto y mapear a IDs de tokens
token_details = map_text_to_token_ids(input_text)

```

```python
# Imprimir los detalles (notar el formateo para mostrar muchos decimales)
print("\nDetalles de los tokens:")
for detail in token_details:
    print(detail)

print("Cantidad de tokens:", len(token_details))
print(f"Costo input prompt: {OpenAI_API_costs['1m']*len(token_details):.10f} $")
```
