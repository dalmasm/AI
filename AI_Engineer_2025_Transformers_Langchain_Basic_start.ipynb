```python
#Abstraccion de LLM de Google (solo de Google, usar init_chat_model para uso generico)
from langchain_google_genai import ChatGoogleGenerativeAI

#El modelo que ponen aca tiene que soportar generacion obviamente
llm = ChatGoogleGenerativeAI(model="gemini-2.0-flash", temperature=1)





```

```python
#Orquestracion basica de componentes
#Prompt, un modelo de AI, y formatear el response standard
# (no todos los modelos tienen el mismo formato)
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
# Definir prompt
prompt = PromptTemplate.from_template("What is the capital of {country}?")

#Definir chain
chain = prompt | llm | StrOutputParser()

# Ejecutar chain
result = chain.invoke({"country": "Germany"})
print(result)
```

```python
# Definir prompt
prompt = PromptTemplate.from_template(
    "Give me three fun facts about {animal} suitable for kids."
)

#Definir chain
chain = prompt | llm | StrOutputParser()

# Ejecutar chain
result = chain.invoke({"animal": "spiders"})
print(result)
```
