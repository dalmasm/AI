```


from transformers import pipeline

```

Para cada tarea NLP, el prompt incluira una palabra clave al principio que servira de instruccion para que el modelo sepa como procesar el prompt. Ejemplo, en el codigo que sigue "summarize" se usa adelante del prompr para marcar la tarea como summarization

Summarization

```
summarizer = pipeline("summarization", model="t5-base")
text = """
The meeting covered quarterly results and future marketing plans.
Revenue was up by 12% compared to last year, driven by an increase in online sales and reduced operational costs.
"""
print(summarizer("summarize: " + text, max_length=50, clean_up_tokenization_spaces=True)[0]['summary_text'])
```

Translation (English to French)

```
translator = pipeline("text2text-generation", model="t5-base")
print(translator("translate English to French: The weather is nice today.", max_length=40)[0]['generated_text'])
```

Question Answering (Text2Text)

```
qa = pipeline("text2text-generation", model="t5-base")
context = "Spain is a country in Europe. Its capital is Madrid. It is known for its food and history."
print(qa(f"question: What is the capital of Spain? context: {context}")[0]['generated_text'])
```

Text Classification (Sentiment)

```
classifier = pipeline("text2text-generation", model="t5-small")
print(classifier("sst2 sentence: I absolutely loved the performance.")[0]['generated_text'])
```

# Instruction Tuning

T5 es el modelo base de Google, entrenado en tasks de NLP generales. Podemos mejorar el concepto de sequence to sequence con un modelo que use como base T5 pero haya sido entrenado para detectar mejores instrucciones incluidas en un prompt,y seguirlas.

Flan T5 es un modelo basado en T5 con Instruction Tuning, acepta lenguaje natural como instrucciones para tareas en vez de keywords especificos.

Few-Shot Prompting

```
from transformers import pipeline

model = pipeline("text2text-generation", model="google/flan-t5-base")
prompt = """
task: sentiment classification
example 1: I loved the book. → positive
example 2: It was boring. → negative
example 3: I had fun watching it. →
"""
print(model(prompt, max_length=10)[0]['generated_text'])
```
